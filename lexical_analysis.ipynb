```python
# getting data
url = 'https://github.com/BobAdamsEE/SouthParkData/blob/master/All-seasons.csv?raw=true'
df = pd.read_csv(url)

# get rid of new lines
df["Line"] = df["Line"].str.replace('\n', '', regex=True)

# make lower case
df["Line"] = df["Line"].str.lower()

# remove punctuation, dealing with missing spaces
def remove_punctuation(text):
    translator = str.maketrans(string.punctuation, ' ' * len(string.punctuation))
    return text.translate(translator)

df['Line'] = df['Line'].apply(remove_punctuation)
df['Line'] = df['Line'].str.replace(r"\s+", r"\s")

# remove fillers
df["Line"] = df["Line"].str.replace('uh', '', regex=True)
df["Line"] = df["Line"].str.replace('um', '', regex=True)
```

```python
# remove stopwords, tokenize and stem
def process_line(text):
    stop_words = set(stopwords.words('english'))
    words = word_tokenize(text)
    filtered_words = [w for w in words if w.lower() not in stop_words]
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(w) for w in filtered_words]
    return ' '.join(stemmed_words)


df["token_line"] = df["Line"].apply(word_tokenize)
df['token_pos'] = df['token_line'].apply(nltk.pos_tag)
df['stemmed_line'] = df['Line'].apply(process_line) # doesn't include stop words
df['stemmed_pos'] = # need to get this by index
```
